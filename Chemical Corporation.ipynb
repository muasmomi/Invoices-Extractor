```python
table2 = read_pdf(input_file,
                  pages="1",
                  silent=True,
                  guess=False,
                  area=(285,42,730,562),
                  columns=[121,331,357,400,440,505,560],
                  pandas_options={'header': None},
                  encoding="windows-1254")
content=table2[0]

content
```

```python

content = content[content[0].notnull()].iloc[:, 0:].reset_index(drop=True)  # Remove NaN
content = content.dropna(subset=[2]).reset_index(drop=True)  # Remove rows with NaN in column 2

content
```

```python
content.rename(columns={
    0: 'Item Code',
    1: 'Description',
    2: 'Quantity',
    3: 'Unit',
    4: 'VAT Code',
    5: 'Unit Price',
    6: 'Net'}, inplace=True)

display(content)
```

```python
dict_content = content.to_dict(orient='records')
dict_content


line_items=[]
for item in dict_content:
    # print(item)
    
    partNum = item['Item Code']
    desc = item['Description']
    quantity = item['Quantity']
    netTotal = item['Net']

    print(partNum)
    
    line_item = {"line_type": "inventory",
                        "sku": partNum,
                        "name": desc,
                        "quantity": int(quantity),
                        "net_total": float(netTotal),
                        "tax_type": "INPUT2"}
    
    line_items.append(line_item)
    
print(line_items)
```
