```python

#To remove "*" and "£"
content[[3,8,9]] = content[[3,8,9]].replace('[£, ]','', regex=True).astype('string')
content[[1]] = content[[1]].replace('[*, ]','', regex=True).astype('string')
content

#print(content)
content = content[content[1].notnull()].iloc[:, 0:].reset_index(drop=True)  # Remove NaN
content = content.dropna(subset=[1,10])  # Remove rows with NaN in column 1,10
content[1] = pd.to_numeric(content[1], errors='coerce')  # Remove anything that not number in column 1

content
```

```python
content.rename(columns={
    0: 'Stock Number',
    1: 'Qty',
    2: 'Description',
    3: 'List Price',
    4: 'Per',
    5: 'Disc 1',
    6: 'Disc 2',
    7: 'Disc 3',
    8: 'Net Price',
    9: 'Value',
    10: "Tax Code ",
    11: "Boolean"}, inplace=True)

display(content)
```

```python
dict_content = content.to_dict(orient='records')
dict_content


line_items=[]
for item in dict_content:
    # print(item)
    
    partNum = item['Stock Number']
    desc = item['Description']
    quantity = item['Qty']
    netTotal = item['Value']

    print(partNum)
    
    line_item = {"line_type": "inventory",
                        "sku": partNum,
                        "name": desc,
                        "quantity": int(quantity),
                        "net_total": float(netTotal),
                        "tax_type": "INPUT2"}
    
    line_items.append(line_item)
    
print(line_items)
```
